# Lab01 - sprint 1: independence of n_center and n_edge
### Independence & Estimations
`n_center` и `n_edge` - числа бракованных чипов в двух зонах одной пластины, поэтому они одной природы по возникновению и по смыслу. $H_0$: величины независимы; $H_1$: связаны. 

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
from pathlib import Path

root = Path.cwd()
if root.name == "notebooks":
    root = root.parent

df = pd.read_parquet(root / "wm811k_features.parquet")
c, e = df["n_center"], df["n_edge"]
N = len(df)
print(df[["n_center", "n_edge"]].describe())

plt.scatter(c, e, s=6, alpha=0.2)
plt.xlabel("n_center, число дефектных чипов в центре")
plt.ylabel("n_edge, число дефектных чипов у края")
plt.title("n_center и n_edge на одной пластине")
plt.show()
```

```python
pr, pr_p = stats.pearsonr(c, e)
sp, sp_p = stats.spearmanr(c, e)
print(f"Пирсон = {pr:.3f} (p = {pr_p:.3g}) | Спирмен = {sp:.3f} (p = {sp_p:.3g})")
```

```python
cb = pd.qcut(c, 5, duplicates="drop")
eb = pd.qcut(e, 5, duplicates="drop")
tab = pd.crosstab(cb, eb)
chi2, p, dof, exp = stats.chi2_contingency(tab)
V = np.sqrt(chi2 / (N * (min(tab.shape) - 1)))
print(tab)
print(f"chi2 = {chi2:.1f}, dof = {dof}, p = {p:.3g}, V Крамера = {V:.3f}, min ожид. = {exp.min():.1f}")
```

```python
rng = np.random.default_rng(0)
rc, re_ = stats.rankdata(c), stats.rankdata(e)
obs = np.corrcoef(rc, re_)[0, 1]
perm = np.array([np.corrcoef(rc, rng.permutation(re_))[0, 1] for _ in range(5000)])
p_perm = (np.sum(np.abs(perm) >= abs(obs)) + 1) / (len(perm) + 1)
print(f"наблюдаемая = {obs:.3f}, p (перестановки) = {p_perm:.4f}")

plt.hist(perm, bins=50)
plt.axvline(obs, color="r", label="наблюдаемая")
plt.xlabel("коэффициент Спирмена при перестановках")
plt.ylabel("число перестановок")
plt.legend()
plt.show()
```

```python
g = df.groupby("lotName")[["n_center", "n_edge"]]
print("между лотами:", g.mean().corr(method="spearman").iloc[0, 1])
dev = df[["n_center", "n_edge"]] - g.transform("mean")
print("внутри лотов:", stats.spearmanr(dev["n_center"], dev["n_edge"])[0])
```

**Диаграмма рассеяния.** Облако точек вытянуто вдоль диагонали: пластины с большим `n_center` в среднем имеют и большее `n_edge`. Разброс при этом большой (среднее `n_center` 22.96 при std 6.12, среднее `n_edge` 41.87 при std 7.80), поэтому связь выражена слабо.

**Корреляция.** Коэффициент Пирсона $r=0.279$ ($p\approx 10^{-282}$), Спирмена $\rho=0.266$ ($p\approx 10^{-254}$). Значения близки, значит, связь близка к линейной и не вызвана выбросами. Коэффициент детерминации $r^2\approx 0.08$: около 8 % дисперсии одной величины «объясняется» другой.

**Хи-квадрат.** Значения разбиты на 5 групп по квантилям, построена таблица сопряжённости 5×5. $\chi^2=1091.2$, число степеней свободы 16, $p\approx 10^{-222}$; минимальная ожидаемая частота 528.4, условие применимости критерия выполнено. Мера силы связи $V=0.131$ (малая). Выборка большая ($N=15881$), поэтому критерий отвергает независимость даже при слабой связи, и вывод о зависимости опирается и на $p$, и на $V$.

**Перестановочный тест.** Мы 5000 раз перемешивали `n_edge` относительно `n_center` и считали $\rho$. При перестановках $\rho$ лежит около 0 (примерно от $-0.03$ до $0.03$), наблюдаемое значение $0.266$ далеко за пределами этого распределения; ни одна перестановка не достигла его ($p<0.0002$). Тест не требует предположений о форме распределения и подтверждает результат корреляционного анализа.

**Источник зависимости.** Корреляция средних по лотам $\rho_{между}=0.648$, корреляция отклонений от среднего своего лота $\rho_{внутри}=0.191$. Значит, значительная часть связи объясняется различиями между лотами: в «плохих» лотах больше брака и в центре, и у края (общие причины на уровне партии, состояния камеры). Но внутри лота связь не исчезает, и это указывает на общие факторы уровня пластины, которые влияют на обе зоны сразу.

**Вывод.** Нулевая гипотеза $H_0$ о независимости `n_center` и `n_edge` отвергнута всеми тремя способами ($\alpha=0.05$). Величины положительно связаны, но связь слабая ($V=0.13$, $\rho\approx 0.27$). Практически это значит, что считать число дефектов в центре и у края независимыми нельзя; при построении модели или генератора, описывающего обе зоны, нужно учитывать их совместное распределение.
